```python
import pandas as pd
from pathlib import Path

def carregar_e_tratar(caminho):
    df = pd.read_csv(
        caminho,
        sep=';',  # Altera o separador (padrão é vírgula ',')
        encoding='utf-8', # Define a codificação dos caracteres (ex: 'latin1' ou 'utf-8')
        header=0, # Define qual linha usar como cabeçalho (0 é a primeira linha)
        decimal=','  # resolve o problema da vírgula decimal em valorMedida, latitude e longitude
    )
    df['datahora'] = pd.to_datetime(df['datahora'])
    return df


RAW = Path('..') / 'dados' / 'raw'

tabelaBA_02 = carregar_e_tratar(RAW / 'BA_Salvador_2022-02.csv')
tabelaBA_10 = carregar_e_tratar(RAW / 'BA_Salvador_2022-10.csv')
tabelaRJ_02 = carregar_e_tratar(RAW / 'RJ_Petropolis_2022-02.csv')
tabelaRJ_07 = carregar_e_tratar(RAW / 'RJ_Petropolis_2022-07.csv')
```

```python
# Etapa para cobertura temporal
# Verificar se os sensores funcionaram durante momentos de crise

# 1. Filtrar as estações de baixa cobertura antes de qualquer análise
estacoes_excluidas = ['Dr. Thouzet - Geo', 'Chácara Flora - Geo', 'São Sebastião - Geo']

tabelaRJ_07_filtrada = tabelaRJ_07[~tabelaRJ_07['nomeEstacao'].isin(estacoes_excluidas)].copy()

print(f"Linhas antes: {len(tabelaRJ_07)} | Linhas depois: {len(tabelaRJ_07_filtrada)}")
```

```python
# 2. Função genérica pra detectar buracos temporais por estação
def detectar_buracos_v2(df, multiplicador=3):
    df = df.sort_values(['codEstacao', 'datahora']).copy()
    df['gap'] = df.groupby('codEstacao')['datahora'].diff()

    # descobre o intervalo TÍPICO de cada estação (a mediana é mais robusta que a média aqui,
    # porque a média seria puxada pelos próprios buracos grandes que queremos detectar)
    intervalo_por_estacao = df.groupby('codEstacao')['gap'].median()
    df['intervalo_esperado'] = df['codEstacao'].map(intervalo_por_estacao)

    # flag: gap muito maior que o intervalo típico daquela estação específica
    buracos = df[df['gap'] > df['intervalo_esperado'] * multiplicador]

    return buracos[['nomeEstacao', 'codEstacao', 'datahora', 'gap', 'intervalo_esperado']]

buracos_RJ_02 = detectar_buracos_v2(tabelaRJ_02)
print(f"Buracos encontrados: {len(buracos_RJ_02)} de {len(tabelaRJ_02)} linhas")
print(buracos_RJ_02.sort_values('gap', ascending=False))
```

```python
# 3. Cruzar com a janela crítica do evento (15/02/2022, ~14h às 20h)
inicio_evento = pd.Timestamp('2022-02-15 14:00:00')
fim_evento = pd.Timestamp('2022-02-15 20:00:00')

buracos_durante_evento = buracos_RJ_02[
    (buracos_RJ_02['datahora'] >= inicio_evento) &
    (buracos_RJ_02['datahora'] <= fim_evento)
]

print("\nBuracos que caíram DENTRO da janela crítica do evento:")
print(buracos_durante_evento)
```

```python
# Função que aplica o MESMO critério de cobertura em qualquer tabela,
# usando a mediana das estações da própria tabela como referência
# (mesma lógica que já usamos nos buracos: comparar cada estação ao padrão do próprio grupo, não a um número fixo assumido)

def estacoes_baixa_cobertura(df, limiar_pct=50):
    contagem = df['nomeEstacao'].value_counts()
    mediana = contagem.median()
    limite = mediana * (limiar_pct / 100)
    return contagem[contagem < limite]

tabelas_brutas = {
    'RJ_02': tabelaRJ_02,
    'RJ_07': tabelaRJ_07,
    'BA_02': tabelaBA_02,
    'BA_10': tabelaBA_10,
}

exclusoes = {}
for nome, df in tabelas_brutas.items():
    baixa_cobertura = estacoes_baixa_cobertura(df)
    exclusoes[nome] = baixa_cobertura.index.tolist()
    print(f"\n{nome} — mediana: {df['nomeEstacao'].value_counts().median():.0f} leituras")
    print(f"Estações excluídas: {exclusoes[nome]}")
    print(baixa_cobertura)

# Aplica a exclusão específica de cada tabela, gerando as versões filtradas definitivas
tabelaRJ_02_filtrada = tabelaRJ_02[~tabelaRJ_02['nomeEstacao'].isin(exclusoes['RJ_02'])].copy()
tabelaRJ_07_filtrada = tabelaRJ_07[~tabelaRJ_07['nomeEstacao'].isin(exclusoes['RJ_07'])].copy()
tabelaBA_02_filtrada = tabelaBA_02[~tabelaBA_02['nomeEstacao'].isin(exclusoes['BA_02'])].copy()
tabelaBA_10_filtrada = tabelaBA_10[~tabelaBA_10['nomeEstacao'].isin(exclusoes['BA_10'])].copy()
```

```python
# Agrupar por dia e somar o volume total de chuva (considerando todas as estações da cidade)
chuva_diaria_BA_02 = tabelaBA_02_filtrada.groupby(
    tabelaBA_02_filtrada['datahora'].dt.date
)['valorMedida'].sum().sort_values(ascending=False)

print("Top 10 dias com mais chuva em Salvador - fevereiro/2022:")
print(chuva_diaria_BA_02.head(10))
```

```python
janelas_criticas = {
    'RJ_02': (pd.Timestamp('2022-02-15 14:00:00'), pd.Timestamp('2022-02-15 20:00:00')),
    'BA_02': (pd.Timestamp('2022-02-14 00:00:00'), pd.Timestamp('2022-02-14 06:00:00')),
}

tabelas_filtradas = {
    'RJ_02': tabelaRJ_02_filtrada,
    'RJ_07': tabelaRJ_07_filtrada,
    'BA_02': tabelaBA_02_filtrada,
    'BA_10': tabelaBA_10_filtrada,
}

resultados_buracos = {}

for nome, df in tabelas_filtradas.items():
    buracos = detectar_buracos_v2(df)
    pct = len(buracos) / len(df) * 100
    resultados_buracos[nome] = buracos

    print(f"\n{nome}: {len(buracos)} buracos de {len(df)} linhas ({pct:.2f}%)")

    if nome in janelas_criticas:
        inicio, fim = janelas_criticas[nome]
        durante_evento = buracos[
            (buracos['datahora'] >= inicio) & (buracos['datahora'] <= fim)
        ]
        print(f"  → {len(durante_evento)} buracos DENTRO da janela crítica ({inicio} a {fim})")
        if len(durante_evento) > 0:
            print(durante_evento)
```

```python
# Recorte largo em UTC: cobre o dia 15 inteiro e a madrugada do dia 16
inicio = pd.Timestamp('2022-02-15 00:00:00')
fim = pd.Timestamp('2022-02-16 06:00:00')

recorte = tabelaRJ_02_filtrada[
    (tabelaRJ_02_filtrada['datahora'] >= inicio) &
    (tabelaRJ_02_filtrada['datahora'] <= fim)
]

# Soma da chuva (todas as estações) por hora de calendário
chuva_por_hora = recorte.groupby(
    recorte['datahora'].dt.floor('h')
)['valorMedida'].sum()

print("Horas com maior volume (UTC):")
print(chuva_por_hora.sort_values(ascending=False).head(10))

print("\nLinha do tempo (UTC):")
print(chuva_por_hora)
```

```python
def janela_critica(df, horas=6):
    # chuva somada (todas as estações) por hora de calendário
    por_hora = df.groupby(df['datahora'].dt.floor('h'))['valorMedida'].sum()

    # completa horas sem leitura com 0, para a janela móvel contar 6 horas de relógio
    por_hora = por_hora.asfreq('h', fill_value=0)

    soma_movel = por_hora.rolling(horas).sum()   # cada valor soma as 6 horas que terminam nele
    ultima_hora = soma_movel.idxmax()            # rótulo da última hora incluída

    inicio = ultima_hora - pd.Timedelta(hours=horas - 1)
    fim = ultima_hora + pd.Timedelta(hours=1)    # fim exclusivo
    return inicio, fim, soma_movel.max()

for nome, df in [('RJ_02', tabelaRJ_02_filtrada), ('BA_02', tabelaBA_02_filtrada)]:
    ini, fim, volume = janela_critica(df)
    print(f"{nome}: {ini} até {fim} (UTC) | volume somado: {volume:.2f}")
```

```python
def buracos_na_janela(buracos, inicio, fim):
    b = buracos.copy()
    b['inicio_gap'] = b['datahora'] - b['gap']
    b['fim_gap'] = b['datahora']

    sobrepoe = (b['inicio_gap'] < fim) & (b['fim_gap'] > inicio)
    b = b[sobrepoe].copy()

    # minutos de silêncio dentro da janela
    ini_ov = b['inicio_gap'].clip(lower=inicio)
    fim_ov = b['fim_gap'].clip(upper=fim)
    b['min_na_janela'] = (fim_ov - ini_ov).dt.total_seconds() / 60
    return b

def estacoes_ativas_na_janela(df, inicio, fim):
    na_janela = df[(df['datahora'] >= inicio) & (df['datahora'] < fim)]
    return na_janela['codEstacao'].nunique(), df['codEstacao'].nunique()

# Exemplo para RJ_02 (repita para BA_02 com os valores novos)
ini_rj, fim_rj, _ = janela_critica(tabelaRJ_02_filtrada)
buracos_rj = detectar_buracos_v2(tabelaRJ_02_filtrada)

dentro = buracos_na_janela(buracos_rj, ini_rj, fim_rj)
print(dentro[['nomeEstacao', 'inicio_gap', 'fim_gap', 'gap', 'min_na_janela']])

ativas, total = estacoes_ativas_na_janela(tabelaRJ_02_filtrada, ini_rj, fim_rj)
print(f"Estações com leitura na janela: {ativas} de {total}")
```

```python
def silencios_iniciados(buracos, df, inicio, fim, min_horas=3):
    b = buracos.copy()
    b['inicio_gap'] = b['datahora'] - b['gap']
    b = b[b['gap'] >= pd.Timedelta(hours=min_horas)]

    na_janela = b[(b['inicio_gap'] >= inicio) & (b['inicio_gap'] < fim)]
    fora = b[(b['inicio_gap'] < inicio) | (b['inicio_gap'] >= fim)]

    horas_janela = (fim - inicio).total_seconds() / 3600
    horas_total = (df['datahora'].max() - df['datahora'].min()).total_seconds() / 3600
    horas_fora = horas_total - horas_janela

    print(f"Silêncios (>= {min_horas} h) iniciados na janela: {len(na_janela)} "
          f"({len(na_janela) / horas_janela:.2f} por hora)")
    print(f"Silêncios iniciados fora da janela: {len(fora)} "
          f"({len(fora) / horas_fora:.3f} por hora)")

silencios_iniciados(buracos_rj, tabelaRJ_02_filtrada, ini_rj, fim_rj)
```

```python
ini_ba, fim_ba, _ = janela_critica(tabelaBA_02_filtrada)
buracos_ba = detectar_buracos_v2(tabelaBA_02_filtrada)

dentro_ba = buracos_na_janela(buracos_ba, ini_ba, fim_ba)
print(dentro_ba[['nomeEstacao', 'inicio_gap', 'fim_gap', 'gap', 'min_na_janela']])

silencios_iniciados(buracos_ba, tabelaBA_02_filtrada, ini_ba, fim_ba)
```
